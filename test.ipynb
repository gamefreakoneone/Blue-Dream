```python
import base64
import mimetypes
from litellm import completion
from dotenv import load_dotenv, find_dotenv

load_dotenv(find_dotenv())

def image_to_data_url(path: str) -> str:
    mime_type, _ = mimetypes.guess_type(path)
    if mime_type is None:
        mime_type = "image/jpeg"
    with open(path, "rb") as f:
        encoded = base64.b64encode(f.read()).decode("utf-8")
    return f"data:{mime_type};base64,{encoded}"

image_data_url = image_to_data_url(r"Storage\screenshots\camera_1\camera_1_2026-01-15_16-27-10.jpg")

resp = completion(
    model="amazon_nova/nova-2-lite-v1",
    messages=[
        {
            "role": "user",
            "content": [
                {"type": "text", "text": "Describe this image in one sentence, then tell me the dominant object."},
                {"type": "image_url", "image_url": {"url": image_data_url}},
            ],
        }
    ],
)

print(resp.choices[0].message.content)
```
